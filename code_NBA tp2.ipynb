# TP 2 : Analyse Probabiliste des Joueurs NBA
## Objectif
Analyser un dataset NBA et calculer des probabilités sur les tailles, âges et poids des joueurs.

```python
# Imports nécessaires
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
```

## 1. Chargement et exploration du dataset NBA

```python
# Charger le dataset NBA (adapter le chemin si nécessaire)
# Dataset Kaggle: https://www.kaggle.com/datasets/justinas/nba-players-data
df = pd.read_csv("nba_all_seasons.csv")

# Aperçu et statistiques descriptives
print(df.head())
print(df.describe())

# Exemple: statistiques sur la taille et l'âge
print(df['player_height'].describe())
print(df['age'].describe())
```

```python
df.columns
```

## 2. Distribution des tailles des joueurs

```python
mean_height = df['player_height'].mean()
std_height = df['player_height'].std()
dist_height = norm(mean_height, std_height)

# Probabilité que la taille soit entre 200 et 210 cm
p_height = dist_height.cdf(210) - dist_height.cdf(200)
print(f"Probabilité qu'un joueur mesure entre 200cm et 210cm: {p_height:.4f}")

# Visualisation
sns.histplot(df['player_height'], bins=30, kde=True)
plt.axvline(mean_height, color='red', linestyle='--')
plt.title("Distribution des tailles des joueurs")
plt.show()
print(std_height)
```


### 1. Statistiques descriptives
L'analyse des statistiques descriptives du dataset révèle des profils bien caractéristiques des athlètes professionnels de haut niveau :
* **Taille (`player_height`) :** La moyenne s'établit à environ **200.56 cm** (soit un peu plus de 2 mètres) avec un écart-type de **9.11 cm**. Les valeurs extrêmes vont de **160.02 cm** (joueurs les plus petits) à **231.14 cm** (joueurs géants comme les pivots). Le premier quartile (Q1) est à 193.04 cm et le troisième quartile (Q3) à 208.28 cm, ce qui montre que la grande majorité des joueurs évolue dans cette fourchette de haute taille, bien au-dessus de la moyenne de la population générale.



## 3. Distribution des âges des joueurs

```python
mean_age = df['age'].mean()
std_age = df['age'].std()
dist_age = norm(mean_age, std_age)

# Probabilité qu'un joueur ait entre 20 et 27 ans
p_age = dist_age.cdf(27) - dist_age.cdf(20)
print(f"Probabilité qu'un joueur ait entre 20 et 27 ans: {p_age:.4f}")

sns.histplot(df['age'], bins=20, kde=True)
plt.axvline(mean_age, color='red', linestyle='--')
plt.title("Distribution des âges des joueurs")
plt.show()
```

### 1. Statistiques descriptives
* **Âge (`age`) :** L'âge moyen des joueurs est de **27.05 ans** avec un écart-type de **4.34 ans**. L'âge minimum enregistré est de **18 ans** (jeunes talents entrés directement en NBA) et l'âge maximum atteint **44 ans** (vétérans d'exception). La médiane (50%) est de 26 ans, ce qui indique une concentration de la ligue sur des athlètes dans la force de l'âge physique.




## 4. Distribution des poids des joueurs

```python
mean_weight = df['player_weight'].mean()
std_weight = df['player_weight'].std()
dist_weight = norm(mean_weight, std_weight)

# Visualisation
sns.histplot(df['player_weight'], bins=30, kde=True)
plt.axvline(mean_weight, color='red', linestyle='--')
plt.title("Distribution des poids des joueurs")
plt.show()
```

### 1. Statistiques descriptives
* **Poids (`player_weight`) :** Le poids moyen est d'environ **100.26 kg** avec un écart-type de **12.43 kg**, avec des valeurs allant de **60.33 kg** à **163.29 kg**, traduisant la diversité morphologique selon les postes sur le terrain (meneurs vs pivots).

### 3. Calculs de probabilités (CDF vs PDF)
Pour quantifier la probabilité d'observer certaines caractéristiques, nous avons utilisé les outils probabilistes de SciPy :
* **Fonction de Densité de Probabilité (PDF - Probability Density Function) :** Elle décrit la *densité* de probabilité autour d'une valeur spécifique. La hauteur de la courbe en un point ne donne pas directement une probabilité (car la probabilité d'une valeur exacte continue est nulle), mais indique où les observations sont les plus probables.
* **Fonction de Répartition (CDF - Cumulative Distribution Function) :** Elle calcule la probabilité cumulée qu'une variable aléatoire soit inférieure ou égale à une valeur donnée ($P(X \le x)$). 
* **Application pratique :** Pour trouver la probabilité qu'un joueur mesure entre 200 cm et 210 cm, nous faisons la soustraction des deux valeurs de la CDF : 
  $$P(200 \le X \le 210) = CDF(210) - CDF(200)$$
  Cette approche nous a permis d'obtenir une probabilité estimée (par exemple, $\approx 37.43\%$ pour la tranche de taille choisie), reflétant fidèlement la proportion réelle des joueurs dans cet intervalle.

### 4. Écarts et valeurs aberrantes (Outliers)
* **Présence d'outliers :** Des valeurs extrêmes apparaissent aux deux extrémités (joueurs extrêmement petits de moins de 170 cm ou très géants au-delà de 225 cm, de même pour les poids lourds > 140 kg).
* **Impact sur la loi normale :** Ces valeurs aberrantes agissent comme du bruit ou des queues de distribution lourdes. Lorsque l'on ajuste une loi normale pure basée uniquement sur la moyenne et l'écart-type globaux, ces outliers peuvent légèrement déformer l'estimation des probabilités, car la loi normale suppose une symétrie et une décroissance rapide des probabilités loin de la moyenne. 
* **Piste d'amélioration :** Modifier l'écart-type (par exemple en le réduisant à $std \times 0.8$) ou filtrer les valeurs aberrantes (méthode de l'écart interquartile - IQR) permet de tester la robustesse du modèle face à ces variations.
